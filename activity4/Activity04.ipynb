```python
from collections import Counter

ciphertext = """
PRCSOFQX FP QDR AFOPQ CZSPR LA JFPALOQSKR.
QDFP FP ZK LIU BROJZK MOLTROE.
"""

letters = [
    c.upper()
    for c in ciphertext
    if c.isalpha()
]

frequency = Counter(letters)

print("Frequency:")
for char, count in frequency.most_common():
    print(char, count)

print("\nTop 3:")
for char, count in frequency.most_common(3):
    print(char, count)
```

```python
def caesar_decrypt(ciphertext, shift):
    result = ""

    for char in ciphertext:
        if char.isalpha():
            base = ord('A') if char.isupper() else ord('a')

            decrypted = chr(
                (ord(char) - base - shift) % 26 + base
            )

            result += decrypted
        else:
            result += char

    return result


ciphertext = "KHOOR ZRUOG"

for shift in range(26):
    plaintext = caesar_decrypt(ciphertext, shift)

    print(f"Shift {shift:2}: {plaintext}")
```
